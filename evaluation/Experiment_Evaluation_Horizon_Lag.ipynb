```python
# Utils: Basis-Funktionen & Helfer
from __future__ import annotations
import os, re, math
from typing import Iterable, List, Tuple, Dict
import numpy as np
import pandas as pd

# (wird in Kachel 2 überschrieben/gefüllt)
ALGORITHM_NAME_MAP: Dict[str, str] = {}

def map_algorithm_name(raw: str) -> str:
    """Nutzt das globale ALGORITHM_NAME_MAP (in Kachel 2 gesetzt)."""
    return ALGORITHM_NAME_MAP.get(str(raw).lower(), str(raw))

def safe_filename(name: str) -> str:
    name = name.strip().replace(" ", "_")
    name = re.sub(r"[^\w.\-]+", "_", name)
    name = re.sub(r"_+", "_", name).strip("_")
    return name

def maybe_log_transform(Z: np.ndarray, base: float = 10.0) -> np.ndarray:
    """Log10-Transform für positive Z; Rest → NaN."""
    Zt = Z.copy().astype(float)
    mask = np.isfinite(Zt) & (Zt > 0)
    Zt[~mask] = np.nan
    Zt[mask] = np.log(Zt[mask]) / math.log(base)
    return Zt

def _camera_eye_from_angles(azimuth_deg: float = 45.0, elevation_deg: float = 45.0, radius: float = 2.6):
    """Azimuth/Elevation (Grad) → Plotly camera.eye (echte 3D-Sicht)."""
    az = math.radians(azimuth_deg)
    el = math.radians(elevation_deg)
    x = radius * math.cos(el) * math.cos(az)
    y = radius * math.cos(el) * math.sin(az)
    z = radius * math.sin(el)
    return dict(x=x, y=y, z=z)

def collect_time_matrix(
    df: pd.DataFrame,
    lags_vals: List[int],
    horizon_vals: List[int],
    value_col: str = "Zeit (ms)",
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Liefert (Z, N):
      Z[h_i, l_j] = Median(value_col) für (horizon=h, lags=l)
      N[h_i, l_j] = Anzahl der Werte
    Pivot ist robust ggü. MultiIndex-Mapping.
    """
    d = df[["lags", "horizon", value_col]].dropna().copy()
    d["lags"] = d["lags"].astype(int)
    d["horizon"] = d["horizon"].astype(int)

    med = d.pivot_table(index="horizon", columns="lags", values=value_col, aggfunc="median")
    cnt = d.pivot_table(index="horizon", columns="lags", values=value_col, aggfunc="count")

    med = med.reindex(index=horizon_vals, columns=lags_vals)
    cnt = cnt.reindex(index=horizon_vals, columns=lags_vals).fillna(0).astype(int)

    return med.values.astype(float), cnt.values.astype(int)
```

```python
# Settings & Style: Mapping, Plotly-Template, Standard-Parameter
import plotly.graph_objects as go
import plotly.io as pio

# Einheitliches Mapping der Modellnamen
ALGORITHM_NAME_MAP = {
    "lstm": "LSTM",
    "gru": "GRU",
    "cnn1d": "1D-CNN",
    "tcn": "TCN",
    "mlp": "MLP",
    "random_forest": "Random Forest",
    "xgboost": "XGBoost",
    "light_xgboost": "XGBoost (light)",
    "lightxgboost": "XGBoost (light)",
    "lightgbm": "LightGBM",
    "catboost": "CatBoost",
    "svm": "SVM",
    "ridge": "Ridge",
}

def apply_global_plotly_style() -> None:
    base_layout = dict(
        font=dict(size=13),
        title=dict(x=0.5, xanchor="center", font=dict(size=20)),
        margin=dict(l=60, r=30, t=60, b=60),
        legend=dict(
            bgcolor="rgba(255,255,255,0.85)",
            bordercolor="rgba(0,0,0,0.15)",
            borderwidth=1,
            font=dict(size=13),
            title=dict(font=dict(size=13)),
            yanchor="top", y=0.98, xanchor="left", x=0.02,
        ),
    )
    scene_axes = dict(
        showbackground=False,
        zeroline=False,
        showspikes=True,
        spikethickness=1,
        spikecolor="rgba(0,0,0,0.35)",
        gridcolor="rgba(0,0,0,0.25)",
        gridwidth=1,
        ticks="outside",
    )
    template = go.layout.Template(
        layout=go.Layout(
            **base_layout,
            scene=dict(
                xaxis=scene_axes | dict(title="Lags"),
                yaxis=scene_axes | dict(title="Horizon"),
                zaxis=scene_axes | dict(title="Zeit (ms)"),
                aspectmode="cube",
            ),
        )
    )
    pio.templates["grid3d_custom"] = template
    pio.templates.default = "grid3d_custom"

apply_global_plotly_style()

# Standard-Parameter (einfach hier anpassen)
OUT_DIR = "plots_3d_grids"
USE_LOG_Z = False
INCLUDE_ZERO_ON_AXES = True         # (0,0) in Draufsicht unten links, Skalen wachsen nach außen
VIEW = "angled_45"                  # "angled_45" (45°) oder "top"
ADD_POINTS = True
SHOW_BASE_PLANE = True

SHOW_COLORBARS = True   # jede Kachel mit eigener Skala
H_SPACING = 0.12        # mehr horizontaler Abstand
V_SPACING = 0.14        # mehr vertikaler Abstand

PANELS_3X3 = [
    ("lstm",         "no-quant", "LSTM — No-Quant"),
    ("lstm",         "quant-16", "LSTM — Quant-16"),
    ("random_forest","no-quant", "Random Forest"),
    ("cnn1d",        "no-quant", "1D-CNN — No-Quant"),
    ("cnn1d",        "quant-16", "1D-CNN — Quant-16"),
    ("xgboost",      "no-quant", "XGBoost"),
    ("light_xgboost","no-quant", "XGBoost (light)"),
    ("svm",          "no-quant", "SVM"),
    ("ridge",        "no-quant", "Ridge"),
]
```

```python
# 3D-Grid (x=Lags, y=Horizon, z=avg_inference_time_ms) – Ansicht wie in der Skizze
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def _zero_axes_traces(xmax: float, ymax: float, z0: float = 0.0):
    """Hilfstraces: sichtbare Achsenlinien bei x=0 und y=0 im z=0-Plane."""
    x_axis = go.Scatter3d(
        x=[0, xmax], y=[0, 0], z=[z0, z0],
        mode="lines", line=dict(width=4), showlegend=False, hoverinfo="skip"
    )
    y_axis = go.Scatter3d(
        x=[0, 0], y=[0, ymax], z=[z0, z0],
        mode="lines", line=dict(width=4), showlegend=False, hoverinfo="skip"
    )
    return x_axis, y_axis

def _camera_like_sketch():
    """Kamera: vorne-links-oben, Blickrichtung zum Zentrum (wie Skizze)."""
    return dict(
        eye=dict(x=-1.9, y=-1.9, z=1.35),     # vorne links, leicht erhöht
        center=dict(x=0, y=0, z=0),
        up=dict(x=0, y=0, z=1),               # z-Achse zeigt nach oben
        projection=dict(type="perspective"),  # natürliche 3D-Wahrnehmung
    )

def plot_3d_grid_surface_time(
    lags_vals: list[int],
    horizon_vals: list[int],
    Z: np.ndarray,
    N: np.ndarray,
    *,
    algorithm_raw: str,
    quant_mode: str,
    use_log_z: bool = USE_LOG_Z,
    out_dir: str = OUT_DIR,
    file_stub: str | None = None,
    include_zero_on_axes: bool = INCLUDE_ZERO_ON_AXES,
    add_points: bool = ADD_POINTS,
    show_base_plane: bool = SHOW_BASE_PLANE,
) -> tuple[str, str | None]:

    os.makedirs(out_dir, exist_ok=True)

    alg_name = map_algorithm_name(algorithm_raw)
    quant_label = {"no-quant":"Nicht quantisiert","quant-16":"Quantisiert (16-Bit)","quant-8":"Quantisiert (8-Bit)"}\
        .get(str(quant_mode).lower(), str(quant_mode))

    # Achsenbereiche: x/y immer ab 0
    x_min = 0 if include_zero_on_axes else min(lags_vals)
    y_min = 0 if include_zero_on_axes else min(horizon_vals)
    x_max = max(lags_vals)
    y_max = max(horizon_vals)

    Z_plot = maybe_log_transform(Z) if use_log_z else Z
    finite_Z = Z_plot[np.isfinite(Z_plot)]
    z_min = 0.0
    z_max = float(np.max(finite_Z)) if finite_Z.size else 1.0

    z_title = "Inference-Zeit (ms)" + (" (log₁₀)" if use_log_z else "")

    fig = go.Figure()

    # Basisebene z=0 (optional)
    if show_base_plane:
        fig.add_trace(go.Surface(
            x=lags_vals, y=horizon_vals, z=np.zeros((len(horizon_vals), len(lags_vals))),
            showscale=False, opacity=0.15, colorscale="Greys", hoverinfo="skip"
        ))

    # 0-Achsen (x=0, y=0)
    xa, ya = _zero_axes_traces(xmax=x_max, ymax=y_max, z0=0.0)
    fig.add_trace(xa)
    fig.add_trace(ya)

    # Surface
    if finite_Z.size:
        fig.add_trace(go.Surface(
            x=lags_vals, y=horizon_vals, z=Z_plot,
            colorscale="Viridis", colorbar=dict(title=z_title),
            opacity=0.97,
            contours=dict(z=dict(show=True, usecolormap=True, highlight=True, project_z=True)),
            hovertemplate=("Lags=%{x}<br>Horizon=%{y}<br>"+z_title+"=%{z:.4f}<br>n=%{customdata}<extra></extra>"),
            customdata=N,
        ))
        if add_points:
            xs, ys, zs, ns = [], [], [], []
            for i, h in enumerate(horizon_vals):
                for j, l in enumerate(lags_vals):
                    zval = Z_plot[i, j]
                    if np.isfinite(zval):
                        xs.append(l); ys.append(h); zs.append(zval); ns.append(N[i, j])
            if xs:
                fig.add_trace(go.Scatter3d(
                    x=xs, y=ys, z=zs, mode="markers",
                    marker=dict(size=3, opacity=0.9),
                    hovertemplate=("Lags=%{x}<br>Horizon=%{y}<br>"+z_title+"=%{z:.4f}<br>n=%{customdata}<extra>Median</extra>"),
                    customdata=ns, showlegend=False,
                ))
    else:
        # Leerer Fall
        fig.add_trace(go.Surface(
            x=lags_vals, y=horizon_vals, z=np.zeros((len(horizon_vals), len(lags_vals))),
            colorscale="Viridis", opacity=0.05, showscale=False, hoverinfo="skip"
        ))
        fig.add_annotation(text="Keine Daten für diese Kombination (alle Werte fehlen).",
                           xref="paper", yref="paper", x=0.5, y=1.02, showarrow=False)

    # Layout – Ursprung vorne links, Achsen wachsen nach außen
    fig.update_layout(
        title=dict(text=f"{alg_name} — {quant_label}<br>"
                        "<sup>Median von avg_inference_time_ms je (lags, horizon)</sup>"),
        scene=dict(
            xaxis=dict(title="Lags", range=[x_min, x_max], showline=True),
            yaxis=dict(title="Horizon (Steps)", range=[y_min, y_max], showline=True),
            zaxis=dict(title=z_title, range=[z_min, z_max], showline=True),
            camera=_camera_like_sketch(),           # deine gewünschte Perspektive
            aspectmode="cube",                      # <<< gleich lange Achsen
            aspectratio=dict(x=1, y=1, z=1),        # <<< explizit 1:1:1
        ),
        margin=dict(l=60, r=30, t=60, b=60),
    )

    if not file_stub:
        file_stub = f"{algorithm_raw}_{quant_mode}_grid_lag_Horizon"
    file_stub = safe_filename(file_stub)
    html_path = os.path.join(out_dir, f"{file_stub}.html")
    fig.write_html(html_path, include_plotlyjs="cdn")

    png_path = None
    try:
        png_path = os.path.join(out_dir, f"{file_stub}.png"); fig.write_image(png_path, scale=2)
    except Exception:
        png_path = None
    return html_path, png_path


# ========= 3×3: 3D-Surfaces je Panel mit gleicher Ansicht wie Skizze =========
def grids_3x3_surfaces_time(
    df: pd.DataFrame,
    *,
    panels: List[Tuple[str, str, str]] = PANELS_3X3,  # aus Kachel 2
    use_log_z: bool = USE_LOG_Z,
    out_dir: str = OUT_DIR,
    file_name: str = "ALL_Algorithms_grid_lag_Horizon_SURFACES_3x3",
    include_zero_on_axes: bool = INCLUDE_ZERO_ON_AXES,
    add_points: bool = False,
    show_base_plane: bool = True,
    horizontal_spacing: float = H_SPACING,
    vertical_spacing: float = V_SPACING,
):
    """
    3×3 Layout: pro Kachel ein 3D-Surface.
    Ursprung (0,0) vorne links, Achsen wachsen nach außen; Skalen je Kachel dynamisch.
    """
    os.makedirs(out_dir, exist_ok=True)
    specs = [[{"type": "scene"} for _ in range(3)] for _ in range(3)]
    fig = make_subplots(rows=3, cols=3, specs=specs,
                        subplot_titles=[t for _, _, t in panels],
                        horizontal_spacing=horizontal_spacing,
                        vertical_spacing=vertical_spacing)

    for idx, (alg, q, _) in enumerate(panels, start=1):
        r = (idx - 1)//3 + 1
        c = (idx - 1)%3 + 1

        dfa = df[df["algorithm"].str.lower() == alg]
        if "quant_mode" in df.columns:
            dfa = dfa[dfa["quant_mode"].str.lower() == q]

        if dfa.empty:
            lags_vals = [0,1]; horizon_vals = [0,1]
            Zp = np.zeros((2,2)); Np = np.zeros_like(Zp, dtype=int)
        else:
            lags_vals = sorted(pd.unique(dfa["lags"].dropna()).astype(int).tolist())
            horizon_vals = sorted(pd.unique(dfa["horizon"].dropna()).astype(int).tolist())
            Z, N = collect_time_matrix(dfa, lags_vals, horizon_vals)
            Zp = maybe_log_transform(Z) if use_log_z else Z; Np = N

        finite_Z = Zp[np.isfinite(Zp)]
        z_min = 0.0
        z_max = float(np.max(finite_Z)) if finite_Z.size else 1.0
        z_title = "Zeit(ms)" + (" (log₁₀)" if use_log_z else "")

        if show_base_plane:
            fig.add_trace(go.Surface(
                x=lags_vals, y=horizon_vals, z=np.zeros((len(horizon_vals), len(lags_vals))),
                showscale=False, opacity=0.15, colorscale="Greys", hoverinfo="skip"
            ), row=r, col=c)

        xa, ya = _zero_axes_traces(xmax=max(lags_vals), ymax=max(horizon_vals), z0=0.0)
        fig.add_trace(xa, row=r, col=c)
        fig.add_trace(ya, row=r, col=c)

        fig.add_trace(go.Surface(
            x=lags_vals, y=horizon_vals, z=Zp,
            colorscale="Viridis", showscale=False,
            opacity=0.97,
            contours=dict(z=dict(show=True, usecolormap=True, highlight=True, project_z=True)),
            hovertemplate=("Lags=%{x}<br>Horizon=%{y}<br>"+z_title+"=%{z:.4f}<br>n=%{customdata}<extra></extra>"),
            customdata=Nb if (Nb:=Np) is not None else None,
        ), row=r, col=c)

        if add_points and finite_Z.size:
            xs, ys, zs, ns = [], [], [], []
            for i, h in enumerate(horizon_vals):
                for j, l in enumerate(lags_vals):
                    zval = Zp[i, j]
                    if np.isfinite(zval):
                        xs.append(l); ys.append(h); zs.append(zval); ns.append(Np[i, j])
            if xs:
                fig.add_trace(go.Scatter3d(
                    x=xs, y=ys, z=zs, mode="markers",
                    marker=dict(size=2.5, opacity=0.85),
                    hovertemplate=("Lags=%{x}<br>Horizon=%{y}<br>"+z_title+"=%{z:.4f}<br>n=%{customdata}<extra>Median</extra>"),
                    customdata=ns, showlegend=False,
                ), row=r, col=c)

        # Szene (je Kachel) – Ursprung vorne links, dynamische Skalen
        scene_key = "" if idx == 1 else str(idx)
        fig.update_layout(**{
            f"scene{scene_key}": dict(
                xaxis=dict(title="Lags", range=[0, max(lags_vals)], showline=True),
                yaxis=dict(title="Horizon (Steps)", range=[0, max(horizon_vals)], showline=True),
                zaxis=dict(title=z_title, range=[z_min, z_max], showline=True),
                camera=_camera_like_sketch(),
                aspectmode="cube",                 # <<< gleich lange Achsen in jeder Kachel
                aspectratio=dict(x=1, y=1, z=1),   # <<< 1:1:1
            )
        })

        if not finite_Z.size:
            fig.add_annotation(text="keine Daten",
                xref=f"x{(r-1)*3+c} domain", yref=f"y{(r-1)*3+c} domain",
                x=0.5, y=0.5, showarrow=False, font=dict(size=12, color="rgba(0,0,0,0.6)")
            )

    fig.update_layout(
        title="Inference-Zeit — 3×3 3D-Grids (Ursprung vorne links, dynamische Skalen)",
        height=1150, width=1250,
        margin=dict(l=50, r=40, t=80, b=60),
    )

    out_stub = safe_filename(file_name)
    html_path = os.path.join(out_dir, f"{out_stub}.html")
    fig.write_html(html_path, include_plotlyjs="cdn")

    png_path = None
    try:
        png_path = os.path.join(out_dir, f"{out_stub}.png"); fig.write_image(png_path, scale=2)
    except Exception:
        png_path = None

    return html_path, png_path

def generate_all_3d_grids_time(
    df: pd.DataFrame,
    *,
    use_log_z: bool = USE_LOG_Z,
    out_dir: str = OUT_DIR,
    only_algorithms: Iterable[str] | None = None,
    only_quant_modes: Iterable[str] | None = None,
    include_zero_on_axes: bool = INCLUDE_ZERO_ON_AXES,
    add_points: bool = ADD_POINTS,
    show_base_plane: bool = SHOW_BASE_PLANE,
) -> List[Tuple[str, str | None]]:
    results: List[Tuple[str, str | None]] = []
    dff = df.copy()

    if only_algorithms is not None:
        dff = dff[dff["algorithm"].str.lower().isin({a.lower() for a in only_algorithms})]
    if "quant_mode" in dff.columns and only_quant_modes is not None:
        dff = dff[dff["quant_mode"].str.lower().isin({q.lower() for q in only_quant_modes})]

    for alg in sorted(dff["algorithm"].dropna().unique(), key=lambda s: str(s).lower()):
        df_alg = dff[dff["algorithm"] == alg]
        quant_modes = (
            sorted(df_alg["quant_mode"].dropna().unique(), key=lambda s: str(s).lower())
            if "quant_mode" in df_alg.columns else ["no-quant"]
        ) or ["no-quant"]

        for q in quant_modes:
            df_aq = df_alg if "quant_mode" not in df_alg.columns else df_alg[df_alg["quant_mode"] == q]
            if df_aq.empty:
                continue

            lags_vals = sorted(pd.unique(df_aq["lags"].dropna()).astype(int).tolist())
            horizon_vals = sorted(pd.unique(df_aq["horizon"].dropna()).astype(int).tolist())
            if not lags_vals or not horizon_vals:
                continue

            Z, N = collect_time_matrix(df_aq, lags_vals, horizon_vals)
            file_stub = f"{alg}_{q}_grid_lag_Horizon"
            paths = plot_3d_grid_surface_time(
                lags_vals, horizon_vals, Z, N,
                algorithm_raw=alg, quant_mode=q,
                use_log_z=use_log_z, out_dir=out_dir, file_stub=file_stub,
                include_zero_on_axes=include_zero_on_axes,
                add_points=add_points, show_base_plane=show_base_plane,
            )
            results.append(paths)
    return results



# --- Beispiel ---
saved_3d  = generate_all_3d_grids_time(df_Grid_edge)
saved_3x3 = grids_3x3_surfaces_time(df_Grid_edge)
print("3D:", len(saved_3d), "  3×3:", saved_3x3[0])
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import numpy as np

def _compute_domains(rows: int, cols: int, hspace: float, vspace: float):
    w = (1.0 - (cols - 1) * hspace) / cols
    h = (1.0 - (rows - 1) * vspace) / rows
    x_domains = []
    for c in range(cols):
        left = c * (w + hspace); right = left + w
        x_domains.append((left, right))
    y_domains = []
    for r in range(rows):  # r=0 ist oberste Reihe
        top = 1.0 - r * (h + vspace); bottom = top - h
        y_domains.append((bottom, top))
    return x_domains, y_domains, w, h


def heatmap_3x3_all_algorithms(
    df: pd.DataFrame,
    *,
    use_log_z: bool = False,
    out_dir: str = "plots_3d_grids",
    file_name: str = "ALL_Algorithms_grid_lag_Horizon_heatmap_3x3",
    include_zero_on_axes: bool = True,
    horizontal_spacing: float = 0.12,
    vertical_spacing: float = 0.14,
    show_colorbars: bool = True,
    panels: List[Tuple[str, str, str]] | None = None,
    subplot_title_size: int = 11,
    axis_title_font_size: int = 11,
    tick_font_size: int = 10,
    colorbar_gap: float = 0.004,          # sehr kleiner Abstand zur Kachelkante
    colorbar_thickness: int = 10,         # schlanke Skala
    right_margin_px: int = 120,           # genug Platz, damit rechte Colorbars außen sichtbar sind
    fig_width: int = 1320,
    fig_height: int = 1000,
):
    panels = panels or PANELS_3X3
    os.makedirs(out_dir, exist_ok=True)

    lg_all = sorted(pd.unique(df["lags"].dropna()).astype(int).tolist())
    hz_all = sorted(pd.unique(df["horizon"].dropna()).astype(int).tolist())
    x0, x1 = (0, max(lg_all)) if include_zero_on_axes else (min(lg_all), max(lg_all))
    y0, y1 = (0, max(hz_all)) if include_zero_on_axes else (min(hz_all), max(hz_all))
    z_title = "Inference-Zeit (ms)" + (" (log₁₀)" if use_log_z else "")

    rows, cols = 3, 3
    fig = make_subplots(
        rows=rows, cols=cols,
        subplot_titles=[t for _, _, t in panels],   # behalten, aber später kleiner setzen
        horizontal_spacing=horizontal_spacing,
        vertical_spacing=vertical_spacing,
    )

    # Domains der Subplots (Paper 0..1)
    x_domains, y_domains, _, _ = _compute_domains(rows, cols, horizontal_spacing, vertical_spacing)

    for idx, (alg, q, _t) in enumerate(panels, start=1):
        r = (idx - 1)//cols + 1
        c = (idx - 1)%cols + 1

        dfa = df[df["algorithm"].str.lower() == alg]
        if "quant_mode" in df.columns:
            dfa = dfa[dfa["quant_mode"].str.lower() == q]

        if dfa.empty:
            Zp = np.full((len(hz_all), len(lg_all)), np.nan)
            Np = np.zeros_like(Zp, dtype=int)
        else:
            Z, N = collect_time_matrix(dfa, lg_all, hz_all)
            Zp = maybe_log_transform(Z) if use_log_z else Z
            Np = N

        finite_vals = Zp[np.isfinite(Zp)]
        if finite_vals.size > 0:
            cmin = float(np.min(finite_vals)); cmax = float(np.max(finite_vals))
        else:
            cmin, cmax = 0.0, 1.0

        coloraxis_name = f"coloraxis{'' if idx == 1 else idx}"

        # === Colorbar-Position EXTERN (für alle Spalten, inkl. rechter Spalte) ===
        xd_left, xd_right = x_domains[c-1]
        yd_bottom, yd_top = y_domains[r-1]
        len_cb = (yd_top - yd_bottom)                  # exakt gleiche Höhe wie Plot
        y_cb   = (yd_bottom + yd_top) / 2.0            # mittig ausrichten

        # Immer rechts außerhalb der Kachel platzieren
        gap = max(colorbar_gap, 0.001)
        x_cb = xd_right + gap                          # auch für rechte Spalte außerhalb
        xanchor = "left"

        fig.update_layout(**{
            coloraxis_name: dict(
                colorscale="Viridis",
                cmin=cmin, cmax=cmax,
                showscale=show_colorbars,
                colorbar=dict(
                    title=dict(text="ms", font=dict(size=tick_font_size)),
                    tickfont=dict(size=tick_font_size),
                    x=x_cb, y=y_cb, len=len_cb, thickness=colorbar_thickness,
                    xanchor=xanchor, yanchor="middle",
                    outlinewidth=0.5,
                ),
            )
        })

        # Heatmap-Trace
        fig.add_trace(
            go.Heatmap(
                x=lg_all, y=hz_all, z=Zp,
                customdata=Np,
                hovertemplate="Lags=%{x}<br>Horizon=%{y}<br>"+z_title+"=%{z:.4f}<br>n=%{customdata}<extra></extra>",
                zsmooth=False,
                coloraxis=coloraxis_name,
            ),
            row=r, col=c
        )

        # Achsen: kleinere Titel, Y = "Horizon"
        fig.update_xaxes(
            title_text="Lags", range=[x0, x1],
            showgrid=True, gridwidth=1, gridcolor="rgba(0,0,0,0.25)",
            title_standoff=10, tickfont=dict(size=tick_font_size),
            title_font=dict(size=axis_title_font_size),
            row=r, col=c
        )
        fig.update_yaxes(
            title_text="Horizon", range=[y0, y1],
            showgrid=True, gridwidth=1, gridcolor="rgba(0,0,0,0.25)",
            title_standoff=10, tickfont=dict(size=tick_font_size),
            title_font=dict(size=axis_title_font_size),
            row=r, col=c
        )

        if finite_vals.size == 0:
            fig.add_annotation(
                text="keine Daten",
                xref=f"x{(r-1)*cols+c} domain", yref=f"y{(r-1)*cols+c} domain",
                x=0.5, y=0.5, showarrow=False, font=dict(size=12, color="rgba(0,0,0,0.6)")
            )

    # Kein Haupttitel; genug rechter Rand, damit externe Colorbars nicht abgeschnitten werden
    fig.update_layout(
        height=fig_height, width=fig_width,
        margin=dict(l=60, r=right_margin_px, t=40, b=60),
        font=dict(size=12),
    )
    # Subplot-Titel kleiner setzen
    if fig.layout.annotations:
        for ann in fig.layout.annotations:
            ann.font = dict(size=subplot_title_size)

    out_stub = safe_filename(file_name)
    html_path = os.path.join(out_dir, f"{out_stub}.html")
    fig.write_html(html_path, include_plotlyjs="cdn")

    png_path = None
    try:
        png_path = os.path.join(out_dir, f"{out_stub}.png")
        fig.write_image(png_path, scale=2)
    except Exception:
        png_path = None
    return html_path, png_path
```

```python
def plot_heatmap_time(
    lags_vals, horizon_vals, Z, N,
    *, algorithm_raw: str, quant_mode: str,
    use_log_z: bool = False, out_dir: str = "plots_3d_grids",
    file_stub: str | None = None, include_zero_on_axes: bool = True,
    axis_title_font_size: int = 11, tick_font_size: int = 10,
):
    import plotly.graph_objects as go
    import os
    os.makedirs(out_dir, exist_ok=True)

    z_title = "Zeit (ms)"
    Z_plot = maybe_log_transform(Z) if use_log_z else Z
    if use_log_z: z_title += " (log₁₀)"

    x0, x1 = (0, max(lags_vals)) if include_zero_on_axes else (min(lags_vals), max(lags_vals))
    y0, y1 = (0, max(horizon_vals)) if include_zero_on_axes else (min(horizon_vals), max(horizon_vals))

    fig = go.Figure(
        data=go.Heatmap(
            x=lags_vals, y=horizon_vals, z=Z_plot,
            colorscale="Viridis",
            colorbar=dict(title=dict(text="ms", font=dict(size=tick_font_size)),
                          tickfont=dict(size=tick_font_size), thickness=12),
            hovertemplate="Lags=%{x}<br>Horizon=%{y}<br>"+z_title+"=%{z:.4f}<br>n=%{customdata}<extra></extra>",
            customdata=N, zsmooth=False,
        )
    )
    # kein Haupttitel; Achsentitel kleiner, Y nur "Horizon"
    fig.update_layout(
        xaxis=dict(
            title=dict(text="Lags", font=dict(size=axis_title_font_size)),
            range=[x0, x1], showgrid=True, gridwidth=1, gridcolor="rgba(0,0,0,0.25)",
            tickfont=dict(size=tick_font_size),
        ),
        yaxis=dict(
            title=dict(text="Horizon", font=dict(size=axis_title_font_size)),
            range=[y0, y1], showgrid=True, gridwidth=1, gridcolor="rgba(0,0,0,0.25)",
            tickfont=dict(size=tick_font_size),
        ),
        margin=dict(l=60, r=30, t=40, b=55),
    )

    if not file_stub:
        file_stub = f"{algorithm_raw}_{quant_mode}_grid_lag_Horizon_heatmap"
    file_stub = safe_filename(file_stub)
    html_path = os.path.join(out_dir, f"{file_stub}.html")
    fig.write_html(html_path, include_plotlyjs="cdn")

    png_path = None
    try:
        png_path = os.path.join(out_dir, f"{file_stub}.png")
        fig.write_image(png_path, scale=2)
    except Exception:
        png_path = None
    return html_path, png_path


def generate_all_heatmaps_time(
    df: pd.DataFrame, *,
    use_log_z: bool = False, out_dir: str = "plots_3d_grids",
    only_algorithms=None, only_quant_modes=None, include_zero_on_axes: bool = True,
    axis_title_font_size: int = 11, tick_font_size: int = 10,
):
    results = []
    dff = df.copy()
    if only_algorithms is not None:
        dff = dff[dff["algorithm"].str.lower().isin({a.lower() for a in only_algorithms})]
    if "quant_mode" in dff.columns and only_quant_modes is not None:
        dff = dff[dff["quant_mode"].str.lower().isin({q.lower() for q in only_quant_modes})]

    for alg in sorted(dff["algorithm"].dropna().unique(), key=lambda s: str(s).lower()):
        df_alg = dff[dff["algorithm"] == alg]
        quant_modes = sorted(df_alg["quant_mode"].dropna().unique(), key=lambda s: str(s).lower()) \
            if "quant_mode" in df_alg.columns else ["no-quant"]
        if not quant_modes:
            quant_modes = ["no-quant"]

        for q in quant_modes:
            df_aq = df_alg if "quant_mode" not in df_alg.columns else df_alg[df_alg["quant_mode"] == q]
            if df_aq.empty:
                continue

            lags_vals = sorted(pd.unique(df_aq["lags"].dropna()).astype(int).tolist())
            horizon_vals = sorted(pd.unique(df_aq["horizon"].dropna()).astype(int).tolist())
            if not lags_vals or not horizon_vals:
                continue

            Z, N = collect_time_matrix(df_aq, lags_vals, horizon_vals)
            file_stub = f"{alg}_{q}_grid_lag_Horizon_heatmap"
            paths = plot_heatmap_time(
                lags_vals, horizon_vals, Z, N,
                algorithm_raw=alg, quant_mode=q,
                use_log_z=use_log_z, out_dir=out_dir,
                file_stub=file_stub, include_zero_on_axes=include_zero_on_axes,
                axis_title_font_size=axis_title_font_size, tick_font_size=tick_font_size,
            )
            results.append(paths)
    return results

# === EINZEL-HEATMAPS (pro algorithm × quant_mode) ==========================
saved_hm = generate_all_heatmaps_time(
    df_Grid_edge,
    use_log_z=False,             # ggf. True bei großen Spannen
    out_dir="plots_3d_grids",    # Ordner wird angelegt
    include_zero_on_axes=True,   # x/y ab 0
    # only_algorithms=["lstm", "cnn1d"],    # optional einschränken
    # only_quant_modes=["no-quant","quant-16"],  # optional einschränken
)

print(f"{len(saved_hm)} Heatmap(s) gespeichert.")
for html_path, png_path in saved_hm:
    print("HTML:", html_path, "| PNG:", png_path)
```

```python
# ============== KACHEL: Model Sizes – Vergleich zweier Konfigurationen ==============
from __future__ import annotations
import os, json, ast
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- Falls Mapping/Helfer nicht vorhanden sind, minimal definieren ---
if "_alg_display" not in globals():
    ALGORITHM_NAME_MAP = {
        "lstm":"LSTM","gru":"GRU","cnn1d":"1D-CNN","cnn_1d":"1D-CNN","cnn":"CNN","tcn":"TCN","mlp":"MLP",
        "random_forest":"Random Forest","rf":"Random Forest","xgboost":"XGBoost",
        "light_xgboost":"XGBoost (light)","lightxgboost":"XGBoost (light)",
        "lightgbm":"LightGBM","catboost":"CatBoost","ridge":"Ridge","ridge_regression":"Ridge",
        "svm":"SVM","svr":"SVR",
    }
    def _alg_display(name: str) -> str:
        if pd.isna(name): return ""
        key = str(name).strip().lower()
        return ALGORITHM_NAME_MAP.get(key, str(name).replace("_"," ").title())

# --- JSON/Value Helfer ---
def _coerce_json_like(val):
    if pd.isna(val): return None
    if isinstance(val, dict): return val
    s = str(val).strip()
    if not s: return None
    try:
        return json.loads(s)
    except Exception:
        try:
            return ast.literal_eval(s)
        except Exception:
            return None

def _to_float(x):
    try:
        f = float(x)
        return f if np.isfinite(f) else np.nan
    except Exception:
        return np.nan

def _summarize_model_size_mb_from_obj(obj) -> float:
    """
    Sucht in einem bereits geparsten JSON-Objekt (dict/list) nach Größenangaben und
    liefert eine einzige Größe in MB zurück (Summe, wenn mehrere Werte vorhanden).
    Unterstützt:
      - 'model_size_mb' (float)
      - 'model_sizes_mb' (dict/list von floats)
      - 'model_size_bytes' / 'model_bytes' (Bytes → MB)
    """
    if obj is None:
        return np.nan
    # Direktwert
    if isinstance(obj, (int, float)):
        val = _to_float(obj)
        if np.isfinite(val): return val
        return np.nan
    if isinstance(obj, list):
        vals = [_summarize_model_size_mb_from_obj(v) for v in obj]
        vals = [v for v in vals if np.isfinite(v)]
        return float(np.nansum(vals)) if vals else np.nan
    if isinstance(obj, dict):
        # priorisierte Schlüssel
        for k in ("model_size_mb",):
            if k in obj:
                v = _to_float(obj.get(k))
                if np.isfinite(v): return v
        for k in ("model_sizes_mb",):
            if k in obj:
                v = obj.get(k)
                if isinstance(v, dict):
                    vals = [_to_float(x) for x in v.values()]
                    vals = [x for x in vals if np.isfinite(x)]
                    if vals: return float(np.sum(vals))
                elif isinstance(v, list):
                    vals = [_to_float(x) for x in v]
                    vals = [x for x in vals if np.isfinite(x)]
                    if vals: return float(np.sum(vals))
        for k in ("model_size_bytes","model_bytes"):
            if k in obj:
                b = _to_float(obj.get(k))
                if np.isfinite(b):
                    return float(b) / (1024*1024)
        # evtl. verschachtelt → rekursiv aufsummieren
        vals = [_summarize_model_size_mb_from_obj(v) for v in obj.values()]
        vals = [v for v in vals if np.isfinite(v)]
        return float(np.nansum(vals)) if vals else np.nan
    return np.nan

def prepare_model_size_df_by_config(df: pd.DataFrame) -> pd.DataFrame:
    """
    Liefert modellgrößen *pro Konfiguration* (algorithm, quant_mode, lags, horizon):
        ['algorithm','quant_mode','lags','horizon','model_size_mb']
    Holt Größen robust aus direkten Spalten oder JSON-Feldern pro Zeile.
    """
    candidates_direct = ["model_size_mb", "model_sizes_mb", "model_size_MB", "modelSizesMB"]
    candidates_json_cols = [
        "error_json", "error_metrics_json", "metrics_json",
        "error_metrics", "error_details_json", "meta_json"
    ]

    rows = []
    for _, row in df.iterrows():
        alg = row.get("algorithm")
        q   = row.get("quant_mode")
        l   = row.get("lags")
        h   = row.get("horizon")

        # Direkte Spalten zuerst
        size_mb = np.nan
        for c in candidates_direct:
            if c in df.columns:
                val = row.get(c)
                if isinstance(val, (list, dict)):
                    size_mb = _summarize_model_size_mb_from_obj(val)
                else:
                    valf = _to_float(val)
                    if np.isfinite(valf):
                        size_mb = valf
                if np.isfinite(size_mb): break

        # Falls nicht gefunden → JSON-Spalten durchsuchen
        if not np.isfinite(size_mb):
            for jc in candidates_json_cols:
                if jc in df.columns:
                    parsed = _coerce_json_like(row.get(jc))
                    size_mb = _summarize_model_size_mb_from_obj(parsed)
                    if np.isfinite(size_mb): break

        rows.append({
            "algorithm": str(alg).lower() if pd.notna(alg) else np.nan,
            "quant_mode": str(q).lower() if pd.notna(q) else np.nan,
            "lags": _to_float(l), "horizon": _to_float(h),
            "model_size_mb": size_mb
        })

    out = pd.DataFrame(rows)
    # Aufräumen
    out = out.dropna(subset=["algorithm","quant_mode","lags","horizon"])
    out["lags"] = out["lags"].astype(int)
    out["horizon"] = out["horizon"].astype(int)
    out["model_size_mb"] = pd.to_numeric(out["model_size_mb"], errors="coerce")

    # nur positive Größen
    out = out[out["model_size_mb"] > 0]
    # Doppelte (mehrere Runs) → Median pro Konfiguration
    out = (out
           .groupby(["algorithm","quant_mode","lags","horizon"], as_index=False)["model_size_mb"]
           .median())
    return out

def plot_model_sizes_for_two_configs(
    df: pd.DataFrame,
    cfg_a: tuple[int,int] = (1,1),
    cfg_b: tuple[int,int] = (28,28),
    *,
    out_dir: str = "plots_3d_grids",
    file_stub: str = "ModelSizes_LagsHorizon_Comparison",
    color_a: str = "#1f77b4",   # Blau
    color_b: str = "#d62728",   # Rot
):
    """
    Vergleicht Modellgrößen (MB) zweier (lags,horizon)-Konfigurationen nebeneinander.
    Zwei Balken je (algorithm × quant_mode): blau (cfg_a) und rot (cfg_b).
    Speichert PNG & PDF im 'out_dir'.
    """
    os.makedirs(out_dir, exist_ok=True)

    sizes = prepare_model_size_df_by_config(df)
    if sizes.empty:
        raise ValueError("Keine Modellgrößen gefunden.")

    la, ha = cfg_a
    lb, hb = cfg_b

    a = sizes[(sizes["lags"]==la) & (sizes["horizon"]==ha)].copy()
    b = sizes[(sizes["lags"]==lb) & (sizes["horizon"]==hb)].copy()
    if a.empty and b.empty:
        raise ValueError(f"Keine Daten für Konfigurationen {cfg_a} und {cfg_b}.")

    # Vollständige Liste der Gruppen (algorithm×quant_mode)
    all_groups = (sizes[["algorithm","quant_mode"]]
                  .drop_duplicates()
                  .sort_values(["algorithm","quant_mode"], kind="stable"))
    all_groups["group_key"] = all_groups["algorithm"] + "||" + all_groups["quant_mode"]

    def to_series(df_cfg):
        df_cfg = df_cfg.copy()
        df_cfg["group_key"] = df_cfg["algorithm"] + "||" + df_cfg["quant_mode"]
        s = df_cfg.set_index("group_key")["model_size_mb"]
        return s

    s_a = to_series(a)
    s_b = to_series(b)

    # Pivot je Gruppe; fehlende → NaN (werden nicht gezeichnet)
    pivot = pd.DataFrame(index=all_groups["group_key"])
    pivot[f"L{la}-H{ha}"] = s_a
    pivot[f"L{lb}-H{hb}"] = s_b

    # Plot
    fig, ax = plt.subplots(figsize=(14, 7))
    x = np.arange(len(pivot.index))
    cols = pivot.columns.tolist()
    n_cols = len(cols)
    bar_width = 0.8 / n_cols

    colors = [color_a, color_b]
    for i, col in enumerate(cols):
        vals = pivot[col].to_numpy(dtype=float)
        ax.bar(
            x + (i - (n_cols - 1)/2)*bar_width,
            vals,
            width=bar_width,
            color=colors[i],
            edgecolor="black",
            linewidth=0.9,
            alpha=0.95,
            label=col
        )

    # Y: log, Limits
    ax.set_yscale("log")
    finite_vals = pivot.to_numpy(dtype=float)
    finite_vals = finite_vals[np.isfinite(finite_vals)]
    if finite_vals.size:
        ymin = finite_vals.min()/1.3
        ax.set_ylim(bottom=max(ymin, 1e-6))

    # Grid & Layout
    ax.grid(axis="y", which="major", linestyle="--", linewidth=1.0, alpha=0.65)
    ax.grid(axis="y", which="minor", linestyle=":", linewidth=0.8, alpha=0.45)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.set_ylabel("Modellgröße (MB, log-Skala)")
    ax.set_xlabel("Modell / Quantisierung")
    ax.set_title(f"Modellgrößen: L{la}-H{ha} (blau) vs. L{lb}-H{hb} (rot)")

    # X-Tick-Labels (mehrzeilig)
    def _tick_label(group_key):
        alg, q = group_key.split("||", 1)
        alg_disp = _alg_display(alg)
        if q == "no-quant":
            return f"{alg_disp}"
        if q == "quant-16":
            return f"{alg_disp}\nQuantisiert\n16 Bit"
        if q == "quant-8":
            return f"{alg_disp}\nQuantisiert\n8 Bit"
        return f"{alg_disp}\n{q}"

    ax.set_xticks(x)
    ax.set_xticklabels([_tick_label(k) for k in pivot.index], ha="center", fontsize=12)
    ax.tick_params(axis="y", which="both", labelsize=12)
    ax.legend(title="Konfiguration", fontsize=12, title_fontsize=12, frameon=True).get_frame().set_alpha(0.95)

    fig.subplots_adjust(bottom=0.28)
    fig.tight_layout()

    # Speichern
    base = os.path.join(out_dir, file_stub)
    png_path = base + ".png"
    pdf_path = base + ".pdf"
    fig.savefig(png_path, dpi=300)
    fig.savefig(pdf_path)
    print(f"Gespeichert: {png_path} und {pdf_path}")
    return fig, ax

# --- Beispielaufruf ---
fig, ax = plot_model_sizes_for_two_configs(df_Grid_edge, cfg_a=(1,1), cfg_b=(28,28), out_dir="plots_3d_grids")
plt.show()
```
